```python
array=[39,56,87,90,21,45,67,12,34,78]
length = len(array)
for i in range(length):
    print("Position:", i, "Value:", array[i])
# This code traverses through the array and prints each element one by one.
```

```python
# sum of all elements in the array
total_sum = 0
for i in range(length):
    total_sum += array[i]
print("Total Sum:", total_sum)
```

```python
# find the maximum element in the array
for i in range(length):
    print ("Checking Position of i:", i, "Value:", array[i])
    if i == 0:
        max_value = array[i]
        print("Initial max_value set to:", max_value)
    elif array[i] > max_value:
        max_value = array[i]
        print("New max_value found:", max_value)
print("Maximum Value:", max_value)
```

```python
# find the minimum element in the array

for i in range(length):
    print ("Checking Position of i:", i, "Value:", array[i])
    if i == 0:
        min_value = array[i]
        print("Initial min_value set to:", min_value)
    elif array[i] < min_value:
        min_value = array[i]
        print("New min_value found:", min_value)
print("Minimum Value:", min_value)
```

```python
# Linear Search for a specific value
please_entered_value = input("Enter a value to search in the array: ")
search_value = int(please_entered_value)
found = False
for i in range(length):
    print("Searching Position of i:", i, "Value:", array[i])
    if array[i] == search_value:
        found = True
        print(f"Value {search_value} found at position {i}.")
        break
if not found:
    print(f"Value {search_value} not found in the array.")  
    
```

```python
class House:
    def __init__(self, address, num_rooms):
        # self.address is a parameter of the instance of the class House
        # address is assigned to the instance variable self.address
        self.address = address
        self.num_rooms = num_rooms

    def display_info(self):
        return f"House is located at {self.address} with {self.num_rooms} rooms."

# Create an instance of the House class
house = House("A704 Hawelia Velenova Park", 3)

# Display the information of the house
print(house.display_info())
```
